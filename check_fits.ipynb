```python
dp_cols = ['OII3726_dp', 'OII3729_dp',
        'Hbeta_dp',
        'OIII4959_dp', 'OIII5007_dp',
        'NII6548_dp', 'Halpha_dp', 'NII6583_dp', 
        'SII6716_dp', 'SII6731_dp']
dp_rank_cols = [f'{col[:-3]}_rank' for col in dp_cols]
classified_catalog_df = Table(classified_catalog[1].data).to_pandas()
classified_catalog_df.drop(columns=dp_cols + dp_rank_cols, inplace=True)
classified_catalog_df['delta_dv'] = classified_catalog_df['dv_r']-classified_catalog_df['dv_l']
```

```python
classified_catalog_df.head(10)
```

```python
classification_map_2comp = {
    2: 'double SF', 8: 'double COMP', 32: 'double AGN', 128: 'double LINER',
    5: 'SF+COMP', 17: 'SF+AGN', 65: 'SF+LINER',
    20: 'COMP+AGN', 68: 'COMP+LINER', 80: 'AGN+LINER',
    257: 'SF+uncertain', 260: 'COMP+uncertain', 272: 'AGN+uncertain', 320: 'LINER+uncertain',
    512: 'unclassified'
}
classified_catalog_df['BPT_class'] = classified_catalog_df['BPT_class'].map(classification_map_2comp)
```
