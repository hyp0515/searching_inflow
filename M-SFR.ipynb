```python
from astropy.io import fits
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from useful_functions import *
```

```python
spectra_data    = fits.open('/Users/hyp0515/data/0715_Spring_BGS_ALL_trimmed.fits')
color_data      = fits.open('/Users/hyp0515/data/0715_Spring_half_BGS_BRIGHT_catalog_with_Flux.fits')
cigale_data     = fits.open('/Users/hyp0515/data/IronPhysProp_v1.2_extracted.fits')
fastspecfit     = fits.open('/Users/hyp0515/data/0715_Spring_half_BGS_BRIGHT_catalog_fastspecfit.fits')
```

```python
SPECTRA = Spectrum(spectra_data, color_data, cigale_data, fastspecfit, load_targetID=None)
logM_all = SPECTRA.df['LOGM'].to_numpy()
logSFR_all = SPECTRA.df['LOGSFR'].to_numpy()


SPECTRA = Spectrum(spectra_data, color_data, cigale_data, fastspecfit, load_targetID=read_ids('mother_samples_ids.txt'))
logM_filtered = SPECTRA.df['LOGM'].to_numpy()
logSFR_filtered = SPECTRA.df['LOGSFR'].to_numpy()
```

```python

plt.plot(logM_all, logSFR_all, 'x', alpha=0.01, color='gray')
# plt.plot(logM_filtered, logSFR_filtered, 'x', alpha=0.05, color='blue')
plt.xlabel('logM')
plt.ylabel('logSFR')
plt.xlim(left=4)
plt.ylim(bottom=-20)
plt.title('Stellar Mass vs Star Formation Rate (ALL)')
plt.savefig('./figures/Jan2/logM_logSFR_all.png', dpi=300)
plt.show()
```

```python
plt.plot(logM_all, logSFR_all, 'x', alpha=0.01, color='gray', label=f'ALL({len(logM_all)})')
plt.plot(logM_filtered, logSFR_filtered, 'x', alpha=0.01, color='blue', label=f'FILTERED({len(logM_filtered)})')
plt.xlabel('logM')
plt.ylabel('logSFR')
plt.xlim(left=4)
plt.ylim(bottom=-20)
plt.title('Stellar Mass vs Star Formation Rate (Filtered)')
plt.legend()
plt.savefig('./figures/Jan2/logM_logSFR_filtered.png', dpi=300)
plt.show()
```

```python
SPECTRA = Spectrum(spectra_data, color_data, cigale_data, fastspecfit, load_targetID=read_ids('dp_candidates_ids.txt'))
logM_dp_cand = SPECTRA.df['LOGM'].to_numpy()
logSFR_dp_cand = SPECTRA.df['LOGSFR'].to_numpy()
```

```python
plt.plot(logM_all, logSFR_all, 'x', alpha=0.01, color='gray', label=f'ALL({len(logM_all)})')
plt.plot(logM_dp_cand, logSFR_dp_cand, 'x', alpha=0.01, color='r', label=f'DP Candidates({len(logM_dp_cand)})')
plt.xlabel('logM')
plt.ylabel('logSFR')
plt.xlim(left=4)
plt.ylim(bottom=-20)
plt.title('Stellar Mass vs Star Formation Rate (DP Candidates)')
plt.legend()
plt.savefig('./figures/Jan2/logM_logSFR_dp_candidates.png', dpi=300)
plt.show()
```

```python
plt.hist(logSFR_all, bins=50, alpha=0.5, label='ALL', color='gray', density=True)
plt.hist(logSFR_dp_cand, bins=50, alpha=0.5, label='DP Candidates', color='r', density=True)
plt.xlabel('logSFR')
plt.ylabel('Density')
plt.xlim(left=-20)
plt.legend()
plt.title('Log Star Formation Rate Distribution (DP Candidates)')
plt.savefig('./figures/Jan2/logSFR_hist_dp_candidates.png', dpi=300)
```
