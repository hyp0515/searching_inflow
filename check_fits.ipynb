```python
# names = [name for name in Table(stellar_fits['EMLINES_MASS'].data).colnames if len(Table(stellar_fits['EMLINES_MASS'].data)[name].shape) <= 1]

stellar_df = Table(stellar_fits['EMLINES_MASS'].data)[['TARGETID', 'MORPHTYPE', 'SERSIC', 'SHAPE_R', 'SHAPE_E1', 'SHAPE_E2', 'VD_SL']].to_pandas()
stellar_df.head(20)
```

```python
b_a_ratio = ((1-np.sqrt(stellar_df['SHAPE_E1']**2 + stellar_df['SHAPE_E2']**2)) /\
    (1+np.sqrt(stellar_df['SHAPE_E1']**2 + stellar_df['SHAPE_E2']**2)))
stellar_df['b_a_ratio'] = b_a_ratio
stellar_df['incl'] = np.arccos(b_a_ratio) * 180 / np.pi
```

```python
plt.hist(stellar_df['b_a_ratio'], bins=30)
plt.show()
```
